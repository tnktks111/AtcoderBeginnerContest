```python
A = [2, 5, 8, 8, 8, 8, 11, 14]
def nibutan(n:int) -> int:
    left, right = 0, len(A) - 1
    while right - left > 1:
        mid = (right + left) // 2
        if A[mid] <= n:
            left = mid
        else:
            right = mid
    while left < len(A) - 1 and A[left] == A[left + 1]:
        left += 1
    return left
```

```python
print(nibutan(-1))
```
