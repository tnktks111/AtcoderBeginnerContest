```python
P = [9, 7, 7, 5, 3, 3]
score_to_rank = {}
for i in range(len(P)):
    if i > 0 and P[i - 1] == P[i]:
        score_to_rank[P[i]] = score_to_rank[P[i - 1]]
    else:
        score_to_rank[P[i]] = i + 1

for s in range(P[-1], P[0]):
    if s in P:
        tmp = score_to_rank[s]
    else:
        score_to_rank[s] = tmp
```

```python
print(score_to_rank)
```
